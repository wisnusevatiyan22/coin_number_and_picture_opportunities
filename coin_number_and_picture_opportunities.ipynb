```python
import random
from collections import Counter
import matplotlib.pyplot as plt

# Fungsi untuk melempar koin
def lempar_koin(jumlah_koin):
    hasil = tuple(random.choice(['A', 'G']) for _ in range(jumlah_koin))
    return hasil

# Fungsi untuk melakukan percobaan
def lakukan_percobaan(jumlah_koin, jumlah_percobaan):
    hasil_percobaan = [lempar_koin(jumlah_koin) for _ in range(jumlah_percobaan)]
    frekuensi = Counter(hasil_percobaan)
    return frekuensi

# Fungsi untuk menghitung probabilitas teoritis
def probabilitas_teoritis(jumlah_koin):
    return {k: 1 / (2 ** jumlah_koin) for k in ['A', 'G']}

# Fungsi untuk menampilkan hasil
def tampilkan_hasil(frekuensi, jumlah_koin, jumlah_percobaan):
    print(f"\nHasil Percobaan {jumlah_koin} Koin ({jumlah_percobaan} Percobaan):")
    for hasil, frek in frekuensi.items():
        print(f"{hasil}: {frek} kali ({(frek/jumlah_percobaan)*100:.2f}%)")

    # Visualisasi
    labels = [''.join(hasil) for hasil in frekuensi.keys()]
    values = list(frekuensi.values())

    plt.figure(figsize=(8, 4))
    plt.bar(labels, values, color='skyblue')
    plt.title(f'Frekuensi Hasil Pelemparan {jumlah_koin} Koin ({jumlah_percobaan} Percobaan)')
    plt.xlabel('Hasil')
    plt.ylabel('Frekuensi')
    plt.show()

# Main program
if __name__ == "__main__":
    # Percobaan 2 koin
    frekuensi_2koin_50 = lakukan_percobaan(2, 50)
    tampilkan_hasil(frekuensi_2koin_50, 2, 50)

    frekuensi_2koin_100 = lakukan_percobaan(2, 100)
    tampilkan_hasil(frekuensi_2koin_100, 2, 100)

    # Percobaan 3 koin
    frekuensi_3koin_50 = lakukan_percobaan(3, 50)
    tampilkan_hasil(frekuensi_3koin_50, 3, 50)

    frekuensi_3koin_100 = lakukan_percobaan(3, 100)
    tampilkan_hasil(frekuensi_3koin_100, 3, 100)

    # Percobaan 4 koin
    frekuensi_4koin_50 = lakukan_percobaan(4, 50)
    tampilkan_hasil(frekuensi_4koin_50, 4, 50)

    frekuensi_4koin_100 = lakukan_percobaan(4, 100)
    tampilkan_hasil(frekuensi_4koin_100, 4, 100)
```

```python
import random
import matplotlib.pyplot as plt

# Fungsi untuk melempar 2 koin
def lempar_2koin():
    return tuple(random.choice(['A', 'G']) for _ in range(2))  # A = Angka, G = Gambar

# Fungsi untuk melakukan percobaan dan menghitung persentase GG
def hitung_persentase_gg(jumlah_percobaan):
    jumlah_gg = 0
    for _ in range(jumlah_percobaan):
        if lempar_2koin() == ('G', 'G'):  # Cek apakah hasilnya GG
            jumlah_gg += 1
    return (jumlah_gg / jumlah_percobaan) * 100  # Persentase GG

# Main program
if __name__ == "__main__":
    # List untuk menyimpan hasil
    percobaan = list(range(1, 10001))  # Percobaan dari 1 hingga 10.000
    persentase_gg = []  # Persentase GG untuk setiap percobaan

    # Lakukan percobaan dari 1 hingga 10.000 kali
    for n in percobaan:
        persentase_gg.append(hitung_persentase_gg(n))

    # Visualisasi hasil
    plt.figure(figsize=(10, 6))
    plt.plot(percobaan, persentase_gg, label="Persentase GG", color="blue")
    plt.axhline(y=25, color="red", linestyle="--", label="Probabilitas Teoritis (25%)")
    plt.title("Persentase GG dalam Pelemparan 2 Koin (1-10.000 Percobaan)")
    plt.xlabel("Jumlah Percobaan")
    plt.ylabel("Persentase GG (%)")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
import random
import matplotlib.pyplot as plt

# Fungsi untuk melempar 2 koin
def lempar_2koin():
    return tuple(random.choice(['A', 'G']) for _ in range(2))  # A = Angka, G = Gambar

# Fungsi untuk menghitung persentase GG
def hitung_persentase_gg(jumlah_percobaan):
    jumlah_gg = 0
    for _ in range(jumlah_percobaan):
        if lempar_2koin() == ('G', 'G'):  # Cek apakah hasilnya GG
            jumlah_gg += 1
    return (jumlah_gg / jumlah_percobaan) * 100  # Persentase GG

# Main program
if __name__ == "__main__":
    # List untuk menyimpan hasil
    percobaan_kumulatif = list(range(500, 10001, 500))  # Percobaan kumulatif setiap 500
    persentase_gg_kumulatif = []  # Persentase GG kumulatif untuk setiap 500 percobaan

    # List untuk grafik real-time
    percobaan_real_time = list(range(1, 10001))  # Percobaan real-time dari 1 hingga 10.000
    persentase_gg_real_time = []  # Persentase GG real-time

    # Lakukan percobaan dari 1 hingga 10.000 kali
    jumlah_gg_total = 0
    for n in percobaan_real_time:
        if lempar_2koin() == ('G', 'G'):
            jumlah_gg_total += 1
        persentase_gg_real_time.append((jumlah_gg_total / n) * 100)

        # Jika mencapai kelipatan 500, simpan hasil kumulatif
        if n in percobaan_kumulatif:
            persentase_gg_kumulatif.append((jumlah_gg_total / n) * 100)
            print(f"Percobaan {n:5} kali: Persentase GG = {persentase_gg_real_time[-1]:.2f}%")

    # Visualisasi grafik real-time
    plt.figure(figsize=(14, 6))

    # Grafik real-time
    plt.subplot(1, 2, 1)
    plt.plot(percobaan_real_time, persentase_gg_real_time, label="Persentase GG Real-Time", color="blue")
    plt.axhline(y=25, color="red", linestyle="--", label="Probabilitas Teoritis (25%)")
    plt.title("Persentase GG Real-Time (1-10.000 Percobaan)")
    plt.xlabel("Jumlah Percobaan")
    plt.ylabel("Persentase GG (%)")
    plt.legend()
    plt.grid(True)

    # Grafik kumulatif setiap 500 percobaan
    plt.subplot(1, 2, 2)
    plt.plot(percobaan_kumulatif, persentase_gg_kumulatif, marker='o', label="Persentase GG Kumulatif", color="green")
    plt.axhline(y=25, color="red", linestyle="--", label="Probabilitas Teoritis (25%)")
    plt.title("Persentase GG Kumulatif (500-10.000 Percobaan)")
    plt.xlabel("Jumlah Percobaan")
    plt.ylabel("Persentase GG (%)")
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```
